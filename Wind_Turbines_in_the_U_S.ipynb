### About Wind Energy

**Wind power** converts wind into electricity using turbines. As wind turns the blades, a rotor spins a generator to produce clean energy—no fuel, no emissions.

In **North America**, especially the **U.S.**, wind is a major renewable energy source. Key facts:
- The U.S. ranks among the top global wind power producers.
- Major wind farms are in **Texas, Iowa, Oklahoma, Kansas, and California**.
- Wind supplies about **10% of U.S. electricity**.
- It helps avoid **329 million metric tons of CO₂ emissions** yearly.

There are two types of wind farms:
- **Onshore** – built on land, easier to access.
- **Offshore** – located in oceans/lakes, higher wind speeds.

Wind power supports **clean energy goals, job growth, and energy independence** across North America.

## Interactive Visualizations with Bokeh

### 1. Let’s see how wind turbine installations have changed over time.

```python
turbines_per_year = df.groupby("Year").size().reset_index(name="Number_of_Turbines")
turbines_per_year.head()
```

```python
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource

output_notebook()

source = ColumnDataSource(turbines_per_year)

p = figure(title="Number of Wind Turbines Installed per Year",
           x_axis_label='Year',
           y_axis_label='Number of Turbines',
           height=350, width=700, tooltips = [('Year', '@Year'), ('Number of Turbines', '@Number_of_Turbines')])

p.line(x='Year', y='Number_of_Turbines', source=source, line_width=2, color="green")
p.scatter(x='Year', y='Number_of_Turbines', source=source, size=5, alpha=0.5)

show(p)
```

In this chart, we’re looking at how many wind turbines were installed each year in the U.S.
Key Insights:
* Before 2000, installations were pretty low and stable — not much activity going on.

* From the early 2000s, there's a clear upward trend. Wind power started gaining traction!

* Around 2008–2012, we see sharp increases, peaking above 6,000 turbines in a single year. That’s a big push toward renewable energy.

* Some dips happened after the peak, but the numbers stayed relatively high compared to the early years.

* Recent years show continued investment, with another big surge around 2020, though it dropped again afterward.

This graph gives us a quick, visual way to understand how wind energy has grown over time — and it shows just how much momentum it gained in the 21st century.

### 2. Turbine Distribution by State

```python
from bokeh.palettes import Viridis256  # Viridis palette with 256 colors
from bokeh.transform import linear_cmap
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource

# Your data setup
state_counts = df['State'].value_counts()
state_names = state_counts.index.astype(str).tolist()

data = {
    'states': state_names,
    'counts': state_counts.values
}

source = ColumnDataSource(data)

# Create the color map using Viridis
color_map = linear_cmap(field_name='counts', palette=Viridis256, low=min(state_counts.values), high=max(state_counts.values))

# Create figure
p2 = figure(x_range=state_names, height=350, title="Top States by Number of Wind Turbines",
            tooltips=[('State', '@states'), ('Number of Turbines', '@counts')])

# Create the bar chart with continuous color mapping using Viridis
p2.vbar(x='states', top='counts', width=0.8, source=source, fill_color=color_map)

# Customize the chart
p2.xgrid.grid_line_color = None
p2.y_range.start = 0
p2.xaxis.major_label_orientation = 1

# Show the plot
show(p2)
```

This bar chart shows how many wind turbines each U.S. state has.

* Texas has the most by far — over 16,000 turbines!
* Iowa and California also have a lot, but there's a big gap after Texas.
* Other states like Oklahoma, Kansas, and Illinois also have many turbines.

Texas leads in wind energy — likely because of its large size, windy areas, and open land.

Wind power is concentrated in a few states with good natural conditions.

There’s a sharp drop after the top states, meaning wind energy is not spread evenly across the country.

### 3. Analyze average turbine capacity by year.

```python
from bokeh.models import ColumnDataSource, CrosshairTool

# Prepare the data
avg_capacity_by_year = df.groupby('Year')['Turbine_Capacity_kW'].mean().reset_index()
source = ColumnDataSource(avg_capacity_by_year)

# Create the plot
p3 = figure(title="Average Turbine Capacity Over Time",
            x_axis_label='Year',
            y_axis_label='Avg Capacity (kW)',
            height=400, width=600,
            tooltips=[('Year', '@Year'), ('Avg Capacity (kW)', '@Turbine_Capacity_kW{0.0}')])

# Line plot
p3.line(x='Year', y='Turbine_Capacity_kW', source=source, line_width=2, color="green")

# Add crosshair tool
p3.add_tools(CrosshairTool())

# Show plot
show(p3)
```

Insight: This chart shows how the average capacity (in kilowatts) of wind turbines installed each year has changed over time.

What it tells us:
* The capacity of turbines has steadily increased since the 1980s.

* Especially after the early 2000s, we see a sharp rise, which likely reflects improvements in turbine technology and a push for more efficient energy generation.

* This suggests that rather than just building more turbines, the U.S. has also focused on building better turbines.

In short: Turbines are getting stronger and smarter over time.

###  5. Wind Turbine Point Map with Bokeh
Requirements:
* Bokeh
* Pandas
* tile_providers for basemap

```python
from bokeh.io import output_file, show
from bokeh.models import ColumnDataSource, HoverTool, WMTSTileSource
from bokeh.plotting import figure
from pyproj import Transformer

# Convert lat/lon to Web Mercator
transformer = Transformer.from_crs("epsg:4326", "epsg:3857")
df["x"], df["y"] = transformer.transform(df["Longitude"].values, df["Latitude"].values)

source = ColumnDataSource(df)

# Set up tile source manually Sets the map layer (tiles) from OpenStreetMap.
#These {Z}/{X}/{Y} values dynamically load tiles depending on zoom and location.

#Bokeh needs a tile provider to display geographical context
tile_url = "http://c.tile.openstreetmap.org/{Z}/{X}/{Y}.png"
tile_source = WMTSTileSource(url=tile_url)

# Set up plot
p5 = figure(title="Wind Turbine Locations", x_axis_type="mercator", y_axis_type="mercator",
           width=900, height=600, tools="pan,wheel_zoom,reset")

p5.add_tile(tile_source)
p5.scatter(x="x", y="y", size=8, fill_color="blue", fill_alpha=0.7, source=source)

hover = HoverTool()
hover.tooltips = [
    ("State", "@State"),
    ("Year", "@Year"),
    ("Capacity (kW)", "@Turbine_Capacity_kW"),
]
p5.add_tools(hover)

output_file("wind_turbine_point_map.html")
show(p5)
```

Insight:
- The map visualizes wind turbine locations across the US.

- The majority of wind turbine installations are concentrated in the central and western parts of the United States, particularly in states like Texas, Oklahoma, Iowa, and Kansas, which are known for their favorable wind conditions and flat terrains.

- The eastern seaboard and Great Lakes region also show notable activity.

- A few scattered points are visible in Hawaii, Caribbean, and Pacific territories, indicating offshore or island-based turbines.

This type of visualization is valuable for policy makers, environmental analysts, and energy companies to identify deployment patterns, evaluate coverage gaps, and plan expansion strategies.

It can also be used to correlate wind energy production with geographic and climatic data.

###  6. Total Wind Turbine Capacity by State - Choropleth Map

```python
#  Create a dictionary to map abbreviations to full names to match json file names
us_state_abbrev_to_name = {
    'AL': 'Alabama', 'AK': 'Alaska', 'AZ': 'Arizona', 'AR': 'Arkansas',
    'CA': 'California', 'CO': 'Colorado', 'CT': 'Connecticut', 'DE': 'Delaware',
    'FL': 'Florida', 'GA': 'Georgia', 'HI': 'Hawaii', 'ID': 'Idaho',
    'IL': 'Illinois', 'IN': 'Indiana', 'IA': 'Iowa', 'KS': 'Kansas',
    'KY': 'Kentucky', 'LA': 'Louisiana', 'ME': 'Maine', 'MD': 'Maryland',
    'MA': 'Massachusetts', 'MI': 'Michigan', 'MN': 'Minnesota', 'MS': 'Mississippi',
    'MO': 'Missouri', 'MT': 'Montana', 'NE': 'Nebraska', 'NV': 'Nevada',
    'NH': 'New Hampshire', 'NJ': 'New Jersey', 'NM': 'New Mexico', 'NY': 'New York',
    'NC': 'North Carolina', 'ND': 'North Dakota', 'OH': 'Ohio', 'OK': 'Oklahoma',
    'OR': 'Oregon', 'PA': 'Pennsylvania', 'RI': 'Rhode Island', 'SC': 'South Carolina',
    'SD': 'South Dakota', 'TN': 'Tennessee', 'TX': 'Texas', 'UT': 'Utah',
    'VT': 'Vermont', 'VA': 'Virginia', 'WA': 'Washington', 'WV': 'West Virginia',
    'WI': 'Wisconsin', 'WY': 'Wyoming'
}

# Assuming your DataFrame is named df
state_capacity = df.groupby('State')['Turbine_Capacity_kW'].sum().reset_index()
state_capacity['State_Name'] = state_capacity['State'].map(us_state_abbrev_to_name)
state_capacity.head()
```

```python
import json
from bokeh.models import GeoJSONDataSource

# Load GeoJSON
with open('us_states.json') as f:
    geojson_data = json.load(f)
```

```python
geo_source = GeoJSONDataSource(geojson=json.dumps(geojson_data))

TOOLTIPS = [
    ('State', '@NAME'),
]
p6 = figure(x_range=(-127, -65), y_range=(23, 50), tooltips=TOOLTIPS, width=1000)
p6.patches('xs', 'ys', fill_color='#2E3440', line_color='#525252', source=geo_source)
show(p6)
```

```python
# Import necessary Bokeh modules and Python libraries
from bokeh.models import GeoJSONDataSource, LinearColorMapper, ColorBar, HoverTool
from bokeh.plotting import figure, show
import json
```

```python
# 1. Calculate percentage of total wind turbine capacity for each row in the dataset
state_capacity["Capacity_Percent"] = (
    state_capacity["Turbine_Capacity_kW"] / state_capacity["Turbine_Capacity_kW"].sum()
) * 100
# This adds a new column showing what % of the total capacity each record represents
state_capacity.head()
```

```python
# 2. Prepare the data for merging with the GeoJSON map
# Group by state to ensure we have one row per state (avoids duplicate index errors later)
grouped_capacity = state_capacity.groupby("State_Name", as_index=False).agg({
    "Turbine_Capacity_kW": "sum",      # Sum total capacity per state
    "Capacity_Percent": "sum"          # Sum total % per state (since grouped rows may add up)
})

# Convert the grouped DataFrame into a dictionary format where each state is a key
# Each key holds another dictionary with capacity and percent
capacity_map = grouped_capacity.set_index("State_Name")[["Turbine_Capacity_kW", "Capacity_Percent"]].to_dict("index")
```

```python
# 3. Merge capacity data into the GeoJSON data used for plotting the map
for feature in geojson_data["features"]:
    state_name = feature["properties"]["NAME"]  # Get the name of the state from GeoJSON
    if state_name in capacity_map:
        # If we have data for this state, attach capacity and percentage values to it
        feature["properties"]["Turbine_Capacity_kW"] = capacity_map[state_name]["Turbine_Capacity_kW"]
        feature["properties"]["Capacity_Percent"] = capacity_map[state_name]["Capacity_Percent"]
    else:
        # If no data is available, fill with zero values
        feature["properties"]["Turbine_Capacity_kW"] = 0
        feature["properties"]["Capacity_Percent"] = 0
```

```python
# 4. Convert the updated GeoJSON data to a Bokeh GeoJSONDataSource
# This allows Bokeh to render geographic shapes from GeoJSON
geo_source = GeoJSONDataSource(geojson=json.dumps(geojson_data))

# 5. Set up a color scale mapper based on the percentage value
# This will color states according to their share of total capacity
color_mapper = LinearColorMapper(palette=Viridis256,
                                 low=state_capacity["Capacity_Percent"].min(),
                                 high=state_capacity["Capacity_Percent"].max())

# 6. Add interactivity with a hover tool
# This shows extra info when you hover over a state
hover = HoverTool(tooltips=[
    ("State", "@NAME"),
    ("Capacity (kW)", "@Turbine_Capacity_kW{0,0}"),
    ("% of Total Capacity", "@Capacity_Percent{0.00}%")
])

# 7. Create the Bokeh figure (canvas for the map)
# Setting ranges to fit the continental US
p7 = figure(title="Wind Turbine Capacity by State (% of Total)",
            x_range=(-127, -65), y_range=(23, 50),
            tools=[hover, "pan", "wheel_zoom", "reset"],  # Enable tools for interaction
            sizing_mode="scale_width", width=1000)        # Responsive width

# 8. Plot the state shapes as patches (polygons) with color mapped to capacity percent
p7.patches("xs", "ys", source=geo_source,
           fill_color={'field': 'Capacity_Percent', 'transform': color_mapper},
           line_color="white", line_width=0.5,
           fill_alpha=1)

# 9. Add a color bar to show what colors represent what % values
color_bar = ColorBar(color_mapper=color_mapper, label_standoff=12,
                     location=(0, 0), title="% of Total Capacity")
p7.add_layout(color_bar, 'right')

# 10. Display the interactive map in the browser
show(p7)
```

This choropleth map visualizes the % total wind turbine capacity by U.S. state. Each state is color-coded based on its capacity, with a gradient scale from dark purple (lower capacity) to bright yellow (highest capacity).

- Moderate levels of capacity are visible in New York, Pennsylvania, and Washington, reflecting investment in renewable infrastructure.
